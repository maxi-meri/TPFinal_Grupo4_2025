```python
#Implementa una funcion que reciba una matriz (lista de listas) y devuelva su transpuesta.

matriz = [
    [1, 2, 3],
    [4, 5, 6],
]

transpuesta = []
for i in range(len(matriz[0])):
    fila = []
    for j in range(len(matriz)):
        fila.append(matriz[j][i])
    transpuesta.append(fila)

print(transpuesta)
```

```python
#Escribe una funci´on que reciba una cadena y devuelva un diccionario con la frecuencia de cada car´acter en la cadena

cadena = input("ingrese una cadena: ")

frecuencia = {}

for caracter in cadena:
    if caracter in frecuencia:
        frecuencia[caracter] += 1
    else:
        frecuencia[caracter] = 1
print(frecuencia)
```

```python
#Implementa una funci´on que reciba dos listas y devuelva una nueva lista con los elementos que aparecen en ambas listas, sin duplicados.

lista1 = [1, 2, 3, 4, 5]
lista2 = [4, 5, 6, 7, 8]

resultado = []

for elemento in lista1:
    if elemento in lista2 and elemento not in resultado:
        resultado.append(elemento)
print(resultado)
```

```python
import numpy as np
import matplotlib.pyplot as plt

lista_registros = []

try:
    with open("registros_clima.txt", "r") as archivo:
      for linea in archivo:
        linea = linea.strip()
        datos = linea.split(",")
        datos = [d.strip() for d in datos]

        id_registro = datos[0]
        ciudad = datos[1]
        temperatura = float(datos[2])
        humedad = int(datos[3])
        estado = datos[4]

        registro = (id_registro, ciudad, temperatura, humedad, estado)
        lista_registros.append(registro)

except FileExistsError:
    print("error critico: Archivo de registros no encontrado")

ciudades_unicas = set(registro[1] for registro in lista_registros)

print("ciudades unicas:")
print(ciudades_unicas)

dic_clima = {}
for registro in lista_registros:
    ciudad = registro[1]
    temperatura = registro[2]
    estado = registro[4]

    if estado == "OK":
        if ciudad not in dic_clima:
            dic_clima[ciudad] = []

        dic_clima[ciudad].append(temperatura)
print("\ndiccionario de clima por ciudad:")
print(dic_clima)

def filtrar_calor (diccionario, ciudad):
    if ciudad in diccionario:
        return list(filter(lambda temp: temp > 30.0, diccionario[ciudad]))
    else:
        return []

print("\ntemperaturas mayores a 30 grados en cada ciudad:")
for ciudad in dic_clima:
    print(f"{ciudad}: {filtrar_calor(dic_clima, ciudad)}")

ciudad_ejemplo = list(dic_clima.keys())[0]
temperaturas = dic_clima[ciudad_ejemplo]

if len(temperaturas) > 5:
    arr = np.array(temperaturas)
    desviacion = np.std(arr)
    print("\ndesviacion estandar:", desviacion)

    x = list(range(len(arr)))
    plt.plot(x, arr)

    plt.title("evolucion de temperaturas en " + ciudad_ejemplo)
    plt.xlabel("tiempo")
    plt.ylabel("temperatura")

    plt.show()
else:
    print("\nno hay suficientes datos para graficar")
```
