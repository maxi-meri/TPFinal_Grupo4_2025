```python
#acumulador

numeros = [1, 2, 3, 4, 5]

suma = 0 #arranca en 0
for n in numeros:
    suma += n #suma = suma + n 
print("La suma es: " + str(suma))
```

```python
#contador

numeros = [1, 2, 3, 4, 5]

contador = 0 #arranca en 0
for n in numeros:
    if n > 3:
        contador += 1 #contador = contador + 1
print("Hay " + str(contador) + " numeros mayores a 3")
```

```python
#maximos y minimos

numeros = [1, 2, 3, 4, 5]

maximo = numeros[0] #arranca con el primer elemento
minimo = numeros[0] #arranca con el primer elemento

for n in numeros:
    if n > maximo:
        maximo = n
    if n < minimo:
        minimo = n

print("El maximo es: " + str(maximo))
print("El minimo es: " + str(minimo))
```

```python
#bandera

numeros = [1, 2, 3, 4, 5]

hay_par = False #arranca en falso
for n in numeros:
    if n % 2 == 0:
        hay_par = True #si encuentra un numero par, cambia la bandera a verdadero
        break #sale del ciclo

print("Hay un numero par: " + str(hay_par))
```

```python
#doble ciclo For (MATRIZ)

matriz = [
    [1, 2, 3],
    [4, 5, 6],
]

suma = 0
for fila in matriz:
    for n in fila:
        suma += n

print("La suma de todos los elementos es: " + str(suma))    
```

```python
#retornar varios valores

def operaciones(a, b):
    suma = a + b
    resta = a - b
    multiplicacion = a * b
    division = a / b if b != 0 else None
    return suma, resta, multiplicacion, division

x, y, z, w = operaciones(10, 5)
print(x, y, z, w) #15, 5, 50, 2.0
```

```python
# Base de datos simulada
estudiantes = {}

# Función para agregar estudiante
def agregar_estudiante(id, nombre, edad):
    # guardamos datos personales en una tupla (inmutable)
    estudiantes[id] = {
        "datos": (nombre, edad),
        "notas": []
    }

# Función para agregar nota
def agregar_nota(id, nota):
    if id in estudiantes:
        estudiantes[id]["notas"].append(nota)
    else:
        print("Estudiante no existe")

# Función para calcular promedio
def calcular_promedio(id):
    if id in estudiantes:
        notas = estudiantes[id]["notas"]
        if len(notas) > 0:
            return sum(notas) / len(notas)
        else:
            return 0
    else:
        return None

# Uso
agregar_estudiante(1, "Juan", 20)
agregar_nota(1, 8)
agregar_nota(1, 6)

print(calcular_promedio(1))  # 7.0
```

```python
#funciones lambda (funciones anonimas)

#funcion normal
def cuadrado(x):
    return x ** 2 

#funcion lambda
cuadrado_lambda = lambda x: x ** 2 #la parte antes de los dos puntos son los parametros, y la parte despues de los dos puntos es el valor que retorna la funcion lambda

print(cuadrado(5)) #25
print(cuadrado_lambda(5)) #25

#con varios parametros
suma = lambda a, b: a + b
print(suma(3, 5)) #8

#con condiciones
par_o_impar = lambda x: "Par" if x % 2 == 0 else "Impar"

print(par_o_impar(4)) #Par
print(par_o_impar(5)) #Impar
```

```python
#string (manipulacion de cadenas)

texto = " Hola Mundo "

#limpiar espacios en blanco
texto_limpio = texto.strip() #elimina los espacios en blanco al inicio y al final del texto
print(texto_limpio) #Hola Mundo

#mayusculas y minusculas
print(texto.upper()) #convierte el texto a mayusculas
print(texto.lower()) #convierte el texto a minusculas

#reemplazar partes del texto
texto_reemplazado = texto.replace("Mundo", "Python") #reemplaza "Mundo" por "Python" en el texto
print(texto_reemplazado) # Hola Python

#dividir un texto en partes
palabras = texto.split() #divide el texto en palabras, usando los espacios como separadores
print(palabras) #['Hola', 'Mundo']

#unir una lista de palabras en un texto
texto_unido = " ".join(palabras) #une las palabras de la lista en un texto, usando un espacio como separador
print(texto_unido) #Hola Mundo
```

```python
#map (trasformar elementos de una lista)

numeros = [1, 2, 3, 4, 5]

#multiplicar cada numero por 2 usando map
numeros_multiplicados = list(map(lambda x: x * 2, numeros))

print(numeros_multiplicados) #[2, 4, 6, 8, 10]
```

```python
#sorted con lambda (ordenar elementos de una lista)

palabras = ["banana", "manzana", "kiwi", "pera"]

#orden normal
print(sorted(palabras)) #['banana', 'kiwi', 'manzana', 'pera']

#orden por longitud de palabra
print(sorted(palabras, key=lambda x: len(x))) #['kiwi', 'pera', 'banana', 'manzana']
```

```python
#list comprehension (otra forma de crear listas)

#forma tradicional
lista = []
for i in range(5):
    lista.append(i * 2) #agrega el numero i multiplicado por 2 a la lista
print(lista) #[0, 2, 4, 6, 8]

#forma con list comprehension
lista_comp = [i * 2 for i in range(5)] #crea una lista con los numeros i multiplicados por 2, para cada i en el rango de 0
print(lista_comp) #[0, 2, 4, 6, 8]

#con condiciones
pares_comp = [i for i in range(10) if i % 2 == 0] #crea una lista con los numeros i, para cada i en el rango de 0 a 9, si i es par
print(pares_comp) #[0, 2, 4, 6, 8]

#con if-else
notas = [7, 8, 5, 6, 9]
resultados = ["Aprobado" if nota >= 6 else "Reprobado" for nota in notas] #crea una lista con "Aprobado" si la nota es mayor o igual a 6, y "Reprobado" si la nota es menor a 6, para cada nota en la lista de notas
print(resultados) #['Aprobado', 'Aprobado', 'Reprobado', 'Aprobado', 'Aprobado']
```

```python
#diccionario comprehension (otra forma de crear diccionarios)

palabras = ["hola", "mundo", "python"]

#palabras : longitud
dicc = {p: len(p) for p in palabras} #crea un diccionario donde la clave es cada palabra de la lista, y el valor es la longitud de esa palabra
print(dicc) #{'hola': 4, 'mundo': 5, 'python': 6}
```

# Decima Unidad

### decorador

*es una funcion que recibe otra funcion, le agrega funcionalidad y devuelve una nueva funcion, sirve para modificar el comportamiento sin cambiar el codigo original*

**Ejemplos de uso:**<br>
*logs*<br>
*manejo de errores*<br>
*validaciones*<br>
*control de acceso*<br>

### sintaxis

*se usa `@` arriba de la funcion*

```python
#decoradores

@decorador #aplica el decorador a la funcion 
def mi_funcion():
    print("Esta es mi funcion") #imprime un mensaje indicando que esta es la funcion original
```

### estructura de un decorador

*un decorador siempre tiene un `wrapper` que es el que "envuelve" la funcion original*

```python
#estructura completa de un decorador

def decorador (func):
    def wrapper(): #esta funcion es la que se ejecuta en lugar de la funcion original
        print("Antes de la funcion original") #codigo que se ejecuta antes de la funcion original
        func() #ejecuta la funcion original
        print("Despues de la funcion original") #codigo que se ejecuta despues de la funcion original
    return wrapper #devuelve la funcion wrapper, que es la que se ejecutara en lugar de la funcion original
```

```python
#ejemplo de decoradores basicos

def logger(func):
    def wrapper():
        print("antes de ejecutar la funcion") #codigo que se ejecuta antes de la funcion original
        func() #ejecuta la funcion original
        print("despues de ejecutar la funcion") #codigo que se ejecuta despues de la funcion original
    return wrapper #devuelve la funcion wrapper, que es la que se ejecutara en lugar de la funcion original

@logger #aplica el decorador logger a la funcion saludo
def saludo():
    print("Hola, bienvenido!") #imprime un mensaje de saludo

saludo() #esto ejecutara la funcion wrapper del decorador logger, que a su vez ejecutara la funcion saludo
```

### decoradores con argumentos

*si la funcion tiene parametros, se usa `*args` y `**kwargs`*

```python
# decoradores con argumentos

def logger(func):
    def wrapper(*args, **kwargs): #la funcion wrapper acepta cualquier numero de argumentos posicionales y de palabras clave
        print("antes") #codigo que se ejecuta antes de la funcion original
        func(*args, **kwargs) #ejecuta la funcion original con sus argumentos
        print("despues") #codigo que se ejecuta despues de la funcion original
    return wrapper #devuelve la funcion wrapper, que es la que se ejecutara en lugar

@logger
def saludo(nombre):
    print(f"Hola, {nombre}!") #imprime un mensaje de saludo personalizado con el nombre proporcionado como argumento 
saludo("Nicolas") #esto ejecutara la funcion wrapper del decorador logger, que a su vez ejecutara la funcion saludo con el argumento "Nicolas"
```

### decoradores anidados

*podes usar varios decoradores, se ejecutan en orden (de abajo hacia arriba)*

```python
# decoradarores anidados 

def decorador1(func):
    def wrapper():
        print("Antes del decorador 1") #codigo que se ejecuta antes de la funcion original
        func() #ejecuta la funcion original
        print("Despues del decorador 1") #codigo que se ejecuta despues de la funcion original
    return wrapper #devuelve la funcion wrapper, que es la que se ejecutara en lugar de la funcion original

def decorador2(func):
    def wrapper():
        print("Antes del decorador 2") #codigo que se ejecuta antes de la funcion original
        func() #ejecuta la funcion original
        print("Despues del decorador 2") #codigo que se ejecuta despues de la funcion original
    return wrapper #devuelve la funcion wrapper, que es la que se ejecutara en lugar de la funcion original

@decorador1 #aplica el decorador1 a la funcion funcion
@decorador2 #aplica el decorador2 a la funcion funcion, que ya esta decorada con decorador1
def funcion(): #imprime un mensaje indicando que esta es la funcion original
    print("Esta es la funcion original")
funcion() #esto ejecutara la funcion wrapper del decorador1, que a su vez ejecutara la funcion wrapper del decorador2, que a su vez ejecutara la funcion original   
```

### decoradores con parametros

*cuando el decorador necesita con una configuracion*

```python
# decoradores con parametros

def repetir(n): #funcion que recibe un parametro n, que indica cuantas veces se debe repetir la ejecucion de la funcion original
    def decorador(func): #funcion que recibe la funcion original como parametro
        def wrapper(*args, **kwargs): #funcion wrapper que acepta cualquier numero de argumentos posicionales y de palabras clave, y se encarga de ejecutar la funcion original n veces 
            for _ in range(n): #repite n veces
                func(*args, **kwargs) #ejecuta la funcion original con sus argumentos
        return wrapper #devuelve la funcion wrapper, que es la que se ejecutara en lugar de la funcion original
    return decorador #devuelve la funcion decorador, que es la que se encargara de decorar la funcion original con el parametro n

@repetir(3) #aplica el decorador repetir con n igual a 3 a la funcion saludo
def saludar():
    print("Hola!") #imprime un mensaje de saludo   
saludar() #esto ejecutara la funcion wrapper del decorador repetir, que a su vez ejecutara la funcion saludo 3 veces
```

```python
#ejemplo general

def logger(func):
    def wrapper(*args, **kwargs):
        print("Ejecutando la funcion...") #codigo que se ejecuta antes de la funcion original
        resultado = func(*args, **kwargs) #ejecuta la funcion original con sus argumentos y guarda el resultado
        print("Funcion ejecutada") #codigo que se ejecuta despues de la funcion original
        return resultado #devuelve el resultado de la funcion original
    return wrapper #devuelve la funcion wrapper, que es la que se ejecutara en lugar de la funcion original 
@logger #aplica el decorador logger a la funcion suma
def suma(a, b):
    return a + b #devuelve la suma de a y b
print(suma(5, 7)) #esto ejecutara la funcion wrapper del decorador logger, que a su vez ejecutara la funcion suma con los argumentos 5 y 7, y luego imprimira el resultado de la suma  
```

### Iteradores

**metodos importantes:**

`__iter__() -> devuelve el iterador, permite que el objeto sea usado en bucles`

`__next__() -> devuelve el siguiente elemento, pero si ya no hay elementos lanza el error "StopIteration"`


```python
#ejemplo de iter y next 

lista = [1, 2, 3, 4] #crea una lista de numeros del 1 al 4

it = iter(lista) #crea un iterador a partir de la lista

print(next(it)) #imprime el siguiente valor del iterador, que sera 1
print(next(it)) #imprime el siguiente valor del iterador, que sera 2    
print(next(it)) #imprime el siguiente valor del iterador, que sera 3
print(next(it)) #imprime el siguiente valor del iterador, que sera 4

print(next(it)) #genera una excepcion StopIteration, porque el iterador ya no tiene mas valores para generar    
```

```python
#el for por dentro

for x in iterable: #iteramos sobre un iterable, como una lista, un conjunto, un diccionario, etc
    print(x) #imprime cada elemento del iterable

#es equivalente a esto

iterador = iter(iterable) #crea un iterador a partir del iterable 

while True: #bucle infinito para iterar sobre el iterador
    try:
        x = next(iterador) #intenta obtener el siguiente valor del iterador
        print(x) #imprime el valor obtenido del iterador
    except StopIteration: #captura la excepcion StopIteration, que se genera cuando el iterador ya no tiene mas valores para generar
        break #sale del bucle infinito cuando se genera la excepcion StopIteration  
```

### generadores: son funciones que utilizan yield

**diferencia entre yield y return**

`return -> termina la funcion`

`yield -> pausa la funcion y guarda el estado para seguir depues` 

*los generadores son utiles para apis, secuencia infintas, archivos enormesy no cargan todo en la ram*

```python
# ejemplo de yield

def mi_gen():
    yield 1 #devuelve el valor 1 y pausa la ejecucion de la funcion
    yield 2 #devuelve el valor 2 y pausa la ejecucion de la funcion
    yield 3 #devuelve el valor 3 y pausa la ejecucion de la funcion

g = mi_gen() #crea un generador a partir de la funcion mi_gen

print(next(g)) #imprime el siguiente valor generado por el generador g, que sera 1
print(next(g)) #imprime el siguiente valor generado por el generador g, que sera 2
print(next(g)) #imprime el siguiente valor generado por el generador g, que sera 3
```

```python
# ejemplo leyendo archivos

def leer_lineas(ruta): #funcion generadora que lee un archivo linea por linea, recibiendo la ruta del archivo como parametro
    with open(ruta, "r") as archivo: #abre el archivo en modo lectura
        for linea in archivo: #itera sobre cada linea del archivo
            yield linea.strip() #devuelve la linea sin espacios en blanco al inicio y al final, y pausa la ejecucion de la funcion hasta que se vuelva a llamar para leer la siguiente linea
```

```python
# ejemplo con generadores infinitos

def contador(): #funcion generadora que cuenta de forma infinita
    n = 1 #inicializa n en 1
    while True: #bucle infinito para contar sin limite
        yield n #devuelve el valor de n y pausa la ejecucion de la funcion hasta que se vuelva a llamar para obtener el siguiente numero
        n += 1 #incrementa n en 1 para contar el siguiente numero
```

```python
#yield from, sirve para delegar parte de la generacion de valores a otro generador

def letras(): #funcion generadora que genera las letras del alfabeto
    yield from "abcde" #delegamos la generacion de valores a la cadena "abcde", que es un iterable que genera cada letra de forma secuencial

#es lo mismo que esto

def letras(): #funcion generadora que genera las letras del alfabeto
    for letra in "abcde": #itera sobre cada letra de la cadena "abcde"
        yield letra #devuelve la letra actual y pausa la ejecucion de la funcion hasta que se vuelva a llamar para generar la siguiente letra
```

### CPU-BOUND vs I/O-BOUND

**CPU-Bound:**<br>
`Poblema -> Mucho calculo matematico`<br>
`Solucion -> Multiprocessing`

**I/O-Bound:**<br>
    `Problema -> Espera red/disco/API`<br>
    `Solucion -> Threading o Asyncio`

### Concurrencia vs Paralelismo

**Concurrencia**<br>
    *hace varias tareas "intercaladas" (no necesariamente al mismo tiempo)*

**Paralelismo**<br>
    *hace varias tareas realmente al mismo tiempo (requiere multiples nucleos)*

### El G.I.L. (Global Interprete Lock)

*Python tiene un "candado" interno llamado el GIL, es decir que python solo puede ejecutar un hilo(threads) a la vez*

**Entonces:**<br>
    *Threads no aceleran los calculos pesados*<br>
    *Pero si sirven para I/O porque le GIL se libera mientras esprea la red/disco*

### Threading

*El Thread es un hilo de ejecucion dentro de un proceso*

**Todos los Threads comparten:**<br>
    *Memoria*<br>
    *Variables*<br>
    *Imports*

```python
# ejemplo de thread

import threading #libreria para crear hilos de ejecucion en paralelo

def worker():
    print("trabajando")

t = threading.Thread(target=worker) #crea un hilo de ejecucion que ejecutara la funcion worker 

t.start() #inicia la ejecucion del hilo, lo que hara que se ejecute la funcion worker en paralelo con el hilo principal del programa
t.join() #espera a que el hilo termine su ejecucion antes de continuar con el resto del programa
```

### cosas importantes

*`taget`: es la funcion a ejecutar, `target=worker` se utiliza sin parentesis ya que si no la ejecuta inmediatamente*

*`start()`: arranca el thread :V*

### Demon Threads

*son hilos descartables, si termina el progrma principal el demon muere automaticamente*

`threading.Thread(target=funcion, daemon=True)`

```python
#ejemplo de demon thread

import threading, time #libreria para crear hilos de ejecucion en paralelo y para manejar el tiempo

def monitoreo_eterno():
    while True: #bucle infinito para monitorear sin parar
        print("ping al servidor ")
        time.sleep(1) #pausa la ejecucion del hilo durante 1 segundo antes de volver a imprimir el mensaje
    
    #demon = true salva al script de colgarse si el hilo de monitoreo_eterno sigue ejecutandose despues de que el hilo principal del programa haya terminado, permitiendo que el programa termine normalmente sin esperar a que el hilo de monitoreo_eterno termine su ejecucion

h = threading.Thread( #crea un hilo de ejecucion que ejecutara la funcion monitoreo_eterno
    target=monitoreo_eterno, #establece la funcion monitoreo_eterno como el objetivo del hilo
    daemon=True ) #establece el hilo como un demonio, lo que significa que se ejecutara en segundo plano y no impedira que el programa termine si el hilo sigue ejecutandose

h.start() #inicia la ejecucion del hilo, lo que hara que se ejecute la funcion monitoreo_eterno en paralelo con el hilo principal del programa
time.sleep(5) #pausa la ejecucion del hilo principal durante 5 segundos, lo que permitira que el hilo de monitoreo_eterno imprima su mensaje varias veces antes de que el hilo principal termine su ejecucion
print("fin, el hilo monitor morira ahora") #imprime un mensaje indicando que el hilo principal ha terminado su ejecucion, lo que hara que el hilo de monitoreo_eterno termine su ejecucion tambien, ya que es un demonio
```

### threadpoolexecutor pool submit

*se combina para procesar las respuestas a medida que llegan en orden cronologico, sin importar el orden inicial*

```python
from concurrent.futures import ThreadPoolExecutor, as_completed #libreria para manejar un pool de hilos de ejecucion en paralelo y para iterar sobre los resultados de las tareas a medida que se van completando
import time #libreria para manejar el tiempo, utilizada para simular tareas que tardan un tiempo en completarse

def tarea(id, pausa): #funcion que simula una tarea que tarda un tiempo en completarse, recibiendo un id para identificar la tarea y un tiempo de pausa para simular la duracion de la tarea
    time.sleep(pausa) #pausa la ejecucion del hilo durante el tiempo simulado, lo que representa el tiempo que tarda la tarea en completarse
    return f"tarea {id} lista" #devuelve un mensaje indicando que la tarea con el id especificado se ha completado

with ThreadPoolExecutor(max_workers=2) as pool: #crea un pool de hilos de ejecucion con un maximo de 2 hilos en paralelo, lo que permite ejecutar hasta 2 tareas al mismo tiempo
    futuros = [
        pool.submit(tarea, id=1, pausa=1), #envia la tarea con id 1 y una pausa de 1 segundo al pool de hilos para que se ejecute en paralelo
        pool.submit(tarea, id=2, pausa=3), #envia la tarea con id 2 y una pausa de 3 segundos al pool de hilos para que se ejecute en paralelo
        pool.submit(tarea, id=3, pausa=2)  #envia la tarea con id 3 y una pausa de 2 segundos al pool de hilos para que se ejecute en paralelo
    ]

    for futuro in as_completed(futuros): #itera sobre los futuros de las tareas a medida que se van completando, lo que permite procesar los resultados de las tareas en el orden en que se completan, en lugar de esperar a que todas las tareas se completen para procesar los resultados
        print(futuro.result()) #imprime el resultado de cada tarea a medida que se va completando, lo que puede mostrar los resultados en un orden diferente al de la lista de tareas debido a la naturaleza paralela de las tareas
```

> ### El Constructor `__init__`

*Es un metodo especial que se ejecuta automaticamente cuando creamos un objeto, esto sirve para inicializar atributos y dejar el objeto listo para usar*

> ### Tambien Usamos `self`

*Es el concepto mas importante al prencipio, `self` representa el objeto actual, es importante ya que debe estar en casi todos los metodos, ademas, de ser el primer parametro que bedemos poner*

`self.nombre = nombre`<br>
*esto quiere decir; el atributo nombre de ESTE objeto*

```python
class Perro:
    def ___init__(self, nombre, edad): #metodo constructor que se ejecuta al crear un objeto de la clase Perro, recibiendo el nombre y la edad del perro como parametros para inicializar los atributos del objeto
        self.nombre = nombre #asigna el valor del parametro nombre al atributo nombre del objeto, lo que permite almacenar el nombre del perro en el objeto
        self.edad = edad #asigna el valor del parametro edad al atributo edad del objeto, lo que permite almacenar la edad del perro en el objeto
```

> ### Metodos

***los metodos son funciones dentro de una clase, permite acceder a atributos y metodos, se usa notacion de puto***

```python
class Perro:
    def __init__(self, nombre):
        self.nombre = nombre

    def ladrar(self):
        print(f"{self.nombre} dice: ¡Guau!") #metodo que imprime un mensaje de ladrido personalizado con el nombre del perro, lo que representa una accion que el perro puede realizar
```

> ### ClassMethod

***Metodos que trabajan sobre la clase completa, usan `cls` en lugar de `self`***

**Cuando usarlo?**<br>
*Cuando el metodo afecta a toda la clase, no a un objeto individual*

```python
class Perro:

    cantidad = 0

    @classmethod #decorador que indica que el metodo siguiente es un metodo de clase, lo que significa que se puede llamar directamente desde la clase sin necesidad de crear un objeto, y que tiene acceso a los atributos de clase
    def total_perros(cls): #metodo de clase que recibe la clase como parametro, lo que permite acceder a los atributos de clase y realizar operaciones relacionadas con la clase en lugar de con objetos individuales
        return cls.cantidad #devuelve el valor del atributo cantidad de la clase, lo que representa el total de perros registrados en la clase
```

> ### StaticMethod

***Son metodos "utilitarios", estna dentro de la clase por organizacion pero no modifican los objetos***

**No Usan**<br>
- *self*<br>
- *cls*

```python
class Matematica:

    @staticmethod #decorador que indica que el metodo siguiente es un metodo estatico, lo que significa que se puede llamar directamente desde la clase sin necesidad de crear un objeto, y que no tiene acceso a los atributos de clase ni a los atributos de instancia
    def sumar(a, b): #metodo estatico que recibe dos numeros como parametros, lo que permite realizar una operacion matematica sin necesidad de acceder a ningun atributo de clase o de instancia
        return a + b #devuelve la suma de los dos numeros recibidos como parametros, lo que representa el resultado de la operacion matematica realizada por el metodo estatico
```

> ### Encapsulamiento

*Consiste en proteger los daots internos del objeto y controlar como se accede y como se modifica*

**Objetivo**<br>
*Es evitar modificaciones incorrectas, estados invalidos y corrupcion de datos*

> ### Niveles de Acceso en Python

*Python usa convenciones, no restricciones estrictas como java*

> ### Publico

*Sin guiones bajos*<br>
`nombre`

```python
#ejemplo Publico
class Persona:
    def __init__(self, documento):
        self.documento = documento #atributo publico

usuario = Persona("12345678") #crea un objeto de la clase Persona con el documento "12345678", lo que representa una persona especifica con su propio documento
print(usuario.documento) #imprime el valor del atributo documento del objeto usuario, que sera "12345678"
usuario.documento = "87654321" #modifica el valor del atributo documento del objeto usuario, lo que representa un cambio en la informacion de la persona
```

> ### Protegido

*Con un guion bajo*<br>
`_nombre`

```python
class Persona:
    def __init__(self, apellido):
        self._apellido = apellido #atributo protegido, se recomienda no acceder directamente desde fuera de la clase, pero no es completamente privado y puede ser accedido si se conoce su nombre
usuario = Persona("Perez") #crea un objeto de la clase Persona con el apellido "Perez", lo que representa una persona especifica con su propio apellido
print(usuario._apellido) #imprime el valor del atributo protegido _apellido del objeto usuario, que sera "Perez", aunque se recomienda no acceder directamente a este atributo desde fuera de la clase        
```

> ### Privado

*Doble guion bajo*<br>
`__nombre`

```python
class Persona:
    def __init__(self, nombre):
        self.__nombre = nombre #atributo privado, no puede ser accedido directamente desde fuera de la clase, lo que garantiza la encapsulacion de la informacion y evita modificaciones no autorizadas del atributo

usuario = Persona("Juan") #crea un objeto de la clase Persona con el nombre "Juan", lo que representa una persona especifica con su propio nombre, lanzara error al intentar acceder al atributo privado __nombre desde fuera de la clase
```

> ### Setters

*Metodos especiales para modificar atributos de forma segura*

```python
#metodo oficial para altera el atributo protegido o privado

def cambiar_nombre(self, nombre_nuevo): #metodo que permite cambiar el valor del atributo privado __nombre de manera controlada, lo que garantiza la encapsulacion de la informacion y evita modificaciones no autorizadas del atributo
    self.__nombre = nombre_nuevo #modifica el valor del atributo privado __nombre con el nuevo nombre proporcionado como parametro, lo que representa un cambio en la informacion de la persona de manera controlada a traves de un metodo de la clase
    print(f"nombre actualizado a {self.__nombre}") #imprime un mensaje indicando que el nombre ha sido actualizado, lo que confirma que el cambio en el atributo privado se ha realizado correctamente a traves del metodo de la clase
```

> ### Herencia

*Permite crear clases hijas a partir de "clases padres", el objetivo de esto es reutilizar codigo*

> ### Superclase

*es la clase general/base se pone entre parentesis*

`super()`<br>
*sirve para acceder a la clase padre*

```python
class AutoDeportivo(Auto): #clase AutoDeportivo que hereda de la clase Auto, lo que significa que AutoDeportivo es una subclase de Auto y hereda sus atributos y metodos, permitiendo crear objetos de AutoDeportivo con las caracteristicas de Auto y agregar funcionalidades adicionales especificas de los autos deportivos

    def __init__(self, marca, modelo, velocidad_maxima): #metodo constructor que se ejecuta al crear un objeto de la clase AutoDeportivo, recibiendo la marca, el modelo y la velocidad maxima del auto deportivo como parametros para inicializar los atributos del objeto
        super().__init__(marca, modelo) #llama al metodo constructor de la clase padre (Auto) para inicializar los atributos heredados de la clase Auto, lo que garantiza que el objeto de AutoDeportivo tenga las caracteristicas basicas de un auto
        self.velocidad_maxima = velocidad_maxima #asigna el valor del parametro velocidad_maxima al atributo velocidad_maxima del objeto, lo que permite almacenar la velocidad maxima del auto deportivo en el objeto y agregar una caracteristica adicional especifica de los autos deportivos
```

> ### Subclase

*Es una clase especializada que puede:*
- *agregar atributos*
- *agregar metodos*
- *modificar comportamiento*

*La clase hija añade sus propios metodos exclusivos, que la clase padre jamas poseyo*

```python
def mostrar_velocidad_maxima(self):
    return f"velocidad punta de {self.velocidad_maxima} km/h" #metodo que devuelve un mensaje indicando la velocidad maxima del auto deportivo, lo que representa una funcionalidad adicional especifica de los autos deportivos que permite obtener informacion sobre su rendimiento

#instaciamos a la hija
mi_ferrari = AutoDeportivo("Ferrari", "F8", "Rojo", 340) #crea un objeto de la clase AutoDeportivo con la marca "Ferrari", el modelo "F8" y una velocidad maxima de 340 km/h, lo que representa un auto deportivo especifico con sus propias caracteristicas y funcionalidades adicionales

print(mi_ferrari.arrancar()) #usa metodo heredado de la clase Auto para arrancar el auto deportivo, lo que demuestra que el objeto de AutoDeportivo tiene las caracteristicas basicas de un auto gracias a la herencia de la clase Auto
print(mi_ferrari.mostrar_velocidad_maxima()) #usa el metodo mostrar_velocidad_maxima para obtener la velocidad maxima del auto deportivo, lo que demuestra la funcionalidad adicional especifica de los autos deportivos que se ha agregado a la clase AutoDeportivo
```

> ### Overriding (Sobrescritura)

*La clase hija redefine un metodo heredado*

```python
class AutoElectrico(Auto): #clase AutoElectrico que hereda de la clase Auto, lo que significa que AutoElectrico es una subclase de Auto y hereda sus atributos y metodos, permitiendo crear objetos de AutoElectrico con las caracteristicas de Auto y agregar funcionalidades adicionales especificas de los autos electricos

    def __init__(self, marca, modelo, color, bateria): #metodo constructor que se ejecuta al crear un objeto de la clase AutoElectrico, recibiendo la marca, el modelo y la autonomia del auto electrico como parametros para inicializar los atributos del objeto
        super().__init__(marca, modelo, color) #llama al metodo constructor de la clase padre (Auto) para inicializar los atributos heredados de la clase Auto, lo que garantiza que el objeto de AutoElectrico tenga las caracteristicas basicas de un auto
        self.bateria = bateria #asigna el valor del parametro bateria al atributo bateria del objeto, lo que permite almacenar la capacidad de la bateria del auto electrico en el objeto y agregar una caracteristica adicional especifica de los autos electricos

    def arrancar(self):
        return f"el {self.marca} arranco silenciosamente {Bateria : {self.bateria}}" #metodo que devuelve un mensaje indicando que el auto electrico ha arrancado de manera silenciosa, lo que representa una funcionalidad adicional especifica de los autos electricos que se diferencia de los autos convencionales y resalta la caracteristica de ser silenciosos al arrancar
    
tesla = AutoElectrico("Tesla", "Model s", "Gris", "100 kwh") #crea un objeto de la clase AutoElectrico con la marca "Tesla", el modelo "Model s", el color "Gris" y una bateria de "100 kwh", lo que representa un auto electrico especifico con sus propias caracteristicas y funcionalidades adicionales

print(tesla.arrancar()) #usa el metodo arrancar para obtener un mensaje indicando que el auto electrico ha arrancado de manera silenciosa, lo que demuestra la funcionalidad adicional especifica de los autos electricos que se ha agregado a la clase AutoElectrico
```
